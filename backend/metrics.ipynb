```python
  ids = sorted(reference_answers.keys())

  references = [reference_answers[qid] for qid in ids]
  candidates = [generated_answers[qid] for qid in ids]

  # Calculate BERTScore in batch
  # model_type can be changed to "distilbert-base-uncased" if roberta-large is too slow/heavy.
  P, R, F1 = bert_score(
      candidates,
      references,
      lang="en",
      model_type="roberta-large",
      rescale_with_baseline=True
  )

  rows = []

  for i, qid in enumerate(ids):
      rouge_scores = calculate_rouge_l(
          reference_answers[qid],
          generated_answers[qid]
      )

      rows.append({
          "id": qid,
          "rougeL_precision": rouge_scores["rougeL_precision"],
          "rougeL_recall": rouge_scores["rougeL_recall"],
          "rougeL_f1": rouge_scores["rougeL_f1"],
          "bertscore_precision": P[i].item(),
          "bertscore_recall": R[i].item(),
          "bertscore_f1": F1[i].item(),
      })

  df = pd.DataFrame(rows)

  # Add average row
  avg_row = {
      "id": "AVERAGE",
      "rougeL_precision": df["rougeL_precision"].mean(),
      "rougeL_recall": df["rougeL_recall"].mean(),
      "rougeL_f1": df["rougeL_f1"].mean(),
      "bertscore_precision": df["bertscore_precision"].mean(),
      "bertscore_recall": df["bertscore_recall"].mean(),
      "bertscore_f1": df["bertscore_f1"].mean(),
  }

  df = pd.concat([df, pd.DataFrame([avg_row])], ignore_index=True)

  print(df.round(4))

  output_file = "automatic_eval_scores.csv"
  df.to_csv(output_file, index=False)

  print(f"\nSaved results to {output_file}")
```
